```python
import os
import math
import pandas as pd
import numpy as np
from datetime import datetime
from IPython.display import display
```

### loading data

```python
data_path = os.path.join("data", "output", "projetos_lattes_1.csv")
df1 = pd.read_csv(data_path, sep=";", encoding='utf-8-sig')
print(df1.shape)
df1.head(3)
```

```python
data_path = os.path.join("data", "output", "projetos_lattes_2.csv")
df2 = pd.read_csv(data_path, sep=";", encoding='utf-8-sig')
print(df2.shape)
df2.head(3)
```

```python
data_path = os.path.join("data", "output", "projetos_lattes_3.csv")
df3 = pd.read_csv(data_path, sep=";", encoding='utf-8-sig')
print(df3.shape)
df3.head(3)
```

```python
df = pd.concat([
    df1, df2, df3
])
print(df.shape)
df.head(3)
```
