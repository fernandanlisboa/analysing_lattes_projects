## Config

### Imports

```python
import os
import math
import pandas as pd
import numpy as np
from datetime import datetime
from IPython.display import display
```

### loading data

```python
data_path = os.path.join("data", "output", "projetos_lattes_completo.csv")
df = pd.read_csv(data_path, sep=";", encoding='utf-8-sig')
print(df.shape)
df.head()
```

## Processing

### Formatting types

```python
df["ano_conclusao"] = df["ano_conclusao"].astype("Int64")
df.info()
```

### Extracting 2025 projects

```python
columns = df.columns.tolist()
print(columns)
```

```python
df["ano_inicio"].unique()
```

```python
df["ano_conclusao"].unique()
```

projetos que nao possuem valor em ano de conclusao significam que ainda estao rodando

```python
current = df.loc[df["ano_conclusao"].isna()]
print(f"{current.shape}\n")
current.head()
```

```python
current["situacao"].unique()
```

#### 2026

```python
df["ano_inicio"].unique()
```

```python
on_going_df = df.loc[df["ano_inicio"] == 2026]
print(f"{on_going_df.shape}\n")
on_going_df.head()
```

```python
finished_this_year = df.loc[df["ano_conclusao"] == 2026]
print(f"{finished_this_year.shape}\n")
finished_this_year.head()
```

```python
this_year_df = pd.concat([
    current,
    on_going_df,
    finished_this_year
])
print(f"{this_year_df.shape}\n")
this_year_df.head()
```

#### 2025

pegando apenas projetos que estiveram em execucao durante 2025 -> isso significa que projetos iniciados ANTES de 2026 e FINALIZADOS ou NAO em 2025

```python
# concluidos em 2025
fb = df.loc[df["ano_conclusao"] == 2025]
print(f"{fb.shape}\n")
fb.head()
```

```python
# em andamento em 2025
og = current.loc[current["ano_inicio"] < 2026]
print(f"{og.shape}\n")
og.head()
```

```python
last_year_df = pd.concat([
    finished_this_year,
    fb,
    og 
])
print(f"{last_year_df.shape}\n")
last_year_df.head(3)
```

```python
last_year_df["ano_inicio"].unique()
```

```python
last_year_df["ano_conclusao"].unique()
```

### Saving

```python
# both
both_df = pd.concat([
    this_year_df,
    last_year_df
]).drop_duplicates().reset_index(drop=True)
print(f"{both_df.shape}\n")
both_df.to_csv(os.path.join("data", "output", "projetos_2025_2026.csv"), sep=";", encoding='utf-8-sig', index=False)
```

```python
this_year_df.to_csv(os.path.join("data", "output", "projetos_2026.csv"), sep=";", encoding='utf-8-sig', index=False)
last_year_df.to_csv(os.path.join("data", "output", "projetos_2025.csv"), sep=";", encoding='utf-8-sig', index=False)
```
